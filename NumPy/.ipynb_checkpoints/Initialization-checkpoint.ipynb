### Different Ways to create NumPy Arrays:

```python
import numpy as np

a=np.array([1,2,3]) # 1D
b=np.array([[1,2,3],[4,5,6]]) # 2D
print(a)
print("-"*50)
print(b)
```

```python
np.arange(1,11,2)
```

```python
np.arange(1,9).reshape(4,2)
```

```python
np.ones((5,5))
```

```python
np.zeros((1,4))
```

```python
np.random.random((3,4))*10
```

```python
np.linspace(-20,20,10)
```

```python
np.identity(5)
```
