#### Indexing and Slicing

```python
import numpy as np
```

```python
a1=np.arange(1,10)
a2=np.random.random((3,4))*100
```

```python
a1
```

```python
# Negative Indexing:
a1[-1]
```

```python
#Slicing in 1D array:
a1[2:6]
```

```python
a2
```

```python
a2[2,1] #(Row,Column)
```

```python
a2[1:,1:3]
```

```python
a2[::2,::3]
```

```python
a2[1,::3]
```

```python
a2[0:2,2:]
```

```python
a3=np.arange(1,28).reshape(3,3,3)
```

```python
a3
```

```python
a3[1,1,1] #(Which 2D array,row,columns)
```

```python
a3[2]
```

```python
a3[0,1,:]
```

```python
a3[::2,::3,::2]
```

```python
a4=np.arange(8).reshape(4,2)
a5=np.arange(12,20).reshape(4,2)
```

```python
a4
```

```python
a5
```

```python
a6=np.hstack((a4,a5))
a6
```

```python
np.hsplit(a6,4)
```

##### Fancy Indexing

```python
a7=np.arange(18).reshape(3,6)
a7
```

```python
a7[[0,1]] # Just past the index inside the list
```

```python
a7[:,[3,4]]
```

##### Boolean Indexing

```python
a8=np.random.randint(1,100,24).reshape(6,4)
a8
```

```python
#Find all numbers greater than 50:
a8[a8>50]
```

```python
#Find all numbers greater than 50 and are even:
a8[(a8>50) & (a8%2==0)]
```

```python
#Find all the numbers not divisble by 7:
a8[~(a8%7==0)]
```
