#### Indexing and Slicing

```python
import numpy as np
```

```python
a1=np.arange(1,10)
a2=np.random.random((3,4))*100
```

```python
a1
```

```python
# Negative Indexing:
a1[-1]
```

```python
#Slicing in 1D array:
a1[2:6]
```

```python
a2
```

```python
a2[2,1] #(Row,Column)
```

```python
a2[1:,1:3]
```

```python
a2[::2,::3]
```

```python
a2[1,::3]
```

```python
a2[0:2,2:]
```

```python
a3=np.arange(1,28).reshape(3,3,3)
```

```python
a3
```

```python
a3[1,1,1] #(Which 2D array,row,columns)
```

```python
a3[2]
```

```python
a3[0,1,:]
```

```python
a3[::2,::3,::2]
```

```python
a4=np.arange(8).reshape(4,2)
a5=np.arange(12,20).reshape(4,2)
```

```python
a4
```

```python
a5
```

```python
a6=np.hstack((a4,a5))
a6
```
