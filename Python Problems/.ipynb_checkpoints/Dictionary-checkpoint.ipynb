### `Q1`: Key with maximum unique values

Given a dictionary with values list, extract key whose value has most unique values.

**Example 1:**

Input:

```bash
test_dict = {"CampusX" : [5, 7, 9, 4, 0], "is" : [6, 7, 4, 3, 3], "Best" : [9, 9, 6, 5, 5]}
```

Output:
```bash
CampusX
```

**Example 2:**

Input:
```bash
test_dict = {"CampusX" : [5, 7, 7, 7, 7], "is" : [6, 7, 7, 7], "Best" : [9, 9, 6, 5, 5]}
```

Output:
```bash
Best
```

````python
test_dict = {"CampusX" : [5, 7, 9, 4, 0], "is" : [6, 7, 4, 3, 3], "Best" : [9, 9, 6, 5, 5]}
````

````python
max_value=0
max_key=''
for (key,value) in test_dict.items():
    if max_value<len(set(value)):
        max_value=len(set(value))
        max_key=key
print(max_key)
    
````

````python
test_dict = {"CampusX" : [5, 7, 7, 7, 7], "is" : [6, 7, 7, 7], "Best" : [9, 9, 6, 5, 5]}
max_value=0
max_key=''
for (key,value) in test_dict.items():
    if max_value<len(set(value)):
        max_value=len(set(value))
        max_key=key
print(max_key)
````

### `Q2`: Replace words from Dictionary. Given String, replace it’s words from lookup dictionary.

**Example 1:**

Input:

```bash
test_str = 'CampusX best for DS students.'
repl_dict = {"best" : "is the best channel", "DS" : "Data-Science"}
```

Output:

```bash
CampusX is the best channel for Data-Science students.
```

**Example 2:**

Input:
```bash
test_str = 'CampusX best for DS students.'
repl_dict = {"good" : "is the best channel", "ds" : "Data-Science"}
```

Output:
```bash
CampusX best for DS students.

````python
test_str = 'CampusX best for DS students.'
repl_dict = {"best" : "is the best channel", "DS" : "Data-Science"}
````

````python
print(test_str.replace("best",repl_dict["best"]).replace("DS",repl_dict["DS"]))
````

````python
intermediate_list=test_str.split()
print(intermediate_list)
````

````python
result=[]
for i in intermediate_list:
    if i not in repl_dict.keys():
        result.append(i)
    else:
        result.append(repl_dict[i])
print(" ".join(result))
````
