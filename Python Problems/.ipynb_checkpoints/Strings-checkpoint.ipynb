```python
# Find the length of the string:
word=input("Enter a word:")
counter=0
for i in word:
    counter+=1
'The length of the string is {}'.format(counter)
```

```python
# Extract the username from email: khadkaarya15@gmail.com
email=input("Enter your email address:")
index_num=email.index("@")
username=email[0:index_num]
print(username)
```

```python
# Count the particular character in a provided string:
word=input("Enter a word:")
search=input("Enter the substring to search:")
counter=0

for i in word:
    if i==search:
        counter+=1
'The character {} appeared {} times in the word {}'.format(search,counter,word)
```

```python
 # Write a program which can remove a particular character from a string:
word=input("Enter a word:")
remove=input("Enter the substring to remove:")

result=''

for i in word:
    if i not in remove:
        result=result+i
print(result)
```
