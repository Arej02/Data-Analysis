```python
# Count the particular character in a provided string:
word=input("Enter a word:")
search=input("Enter the substring to search:")
counter=0

for i in word:
    if i==search:
        counter+=1
'The character {} appeared {} times in the word {}'.format(search,counter,word)
```

```python
 # Write a program which can remove a particular character from a string:
word=input("Enter a word:")
remove=input("Enter the substring to remove:")

result=''

for i in word:
    if i not in remove:
        result=result+i
print(result)
```

```python
# Write a program to check if the string is palindrome or not:
# Method 1:
word=input("Enter a word:")
reverse_word=word[::-1]
if word==reverse_word:
    print("Palindrome String")
else:
    print("Not palindrome")
```

```python
# Method 2:
word=input("Enter a word:")
flag=True
for i in range(len(word)//2):
    if word[i]!=word[len(word)-i-1]:
        flag=False
        print("Not Palindrome")
        break

if flag:
    print("Palindrome")
    
```

```python
# Write a program to count the number of words in a string:
sentence=input("Enter a sentence:")
temp_string=''
temp_list=[]

for i in sentence:
    if i!=' ':
        temp_string+=i
    else:
        temp_list.append(temp_string)
        temp_string=''
temp_list.append(temp_string)
print(len(temp_list))
```
