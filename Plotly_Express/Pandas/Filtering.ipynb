```python
# Find all the finals winners:
df[df['MatchNumber']=='Final'][["Season","WinningTeam"]]
```

```python
# How many super overs finsihes have occured:
df[df['SuperOver']=="Y"].shape[0]
```

```python
# How many matches has CSK won in Kolkata:
df[(df['City']=="Kolkata") & (df["WinningTeam"]=="Chennai Super Kings")].shape[0]
```

```python
# Toss winner is the match winner:
toss_win=df[df["TossWinner"]==df["WinningTeam"]].shape[0]
(toss_win/df.shape[0])*100
```

```python
# Display the bar chart of top 5 teams who have played maximum number of matches in the IPL.
(df['Team1'].value_counts()+df['Team2'].value_counts()).sort_values().head(5).plot(kind='barh')
```

```python
#  Player who got Most no. of player of the match award playing against Mumbai Indians.
# Just for this question assume player of the match award is given to players from winning team. Although this is true in most of the cases.
mask1=df[(df['Team1']=="Mumbai Indians") | (df['Team2']=="Mumbai Indians")]
mask1[mask1["WinningTeam"] != "Mumbai Indians"]['Player_of_Match'].value_counts().sort_values(ascending=False).head(1)
```

```python
# Create a function which will take two string(name of two teams) as input. Show win Loss record between them and player getting most player of the match award in matches between these two teams.
def dashboard(team1, team2):
    mask = (
        ((df['Team1'] == team1) & (df['Team2'] == team2)) |
        ((df['Team1'] == team2) & (df['Team2'] == team1))
    )

    print(df[mask]['WinningTeam'].value_counts())
    print(df[mask]['Player_of_Match'].value_counts().head(1))
```

```python
dashboard('Kolkata Knight Riders','Chennai Super Kings')
```

```python
# Find out the top 7 cities where the matches of Kolkata Knight Riders are played frequently and plot the result as bar chart. 
df[(df['Team1'] == 'Kolkata Knight Riders') | (df['Team2'] == 'Kolkata Knight Riders')]['City'].value_counts().head(7).plot(kind='bar')
```

```python
# Find out the average margin for the team Mumbai Indians for only the season 2011.
df[((df['Team1'] == 'Mumbai Indians') | (df['Team2'] == 'Mumbai Indians')) 
    & (df['Season']=='2011')]['Margin'].mean()
```
