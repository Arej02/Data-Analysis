```python
import pandas as pd
import numpy as np
```

```python
df=pd.read_csv("ipl-matches.csv")
df.shape
```

```python
df.head()
```

```python
# Find all the finals winners:
df[df['MatchNumber']=='Final'][["Season","WinningTeam"]]
```

```python
# How many super overs finsihes have occured:
df[df['SuperOver']=="Y"].shape[0]
```

```python
# How many matches has CSK won in Kolkata:
df[(df['City']=="Kolkata") & (df["WinningTeam"]=="Chennai Super Kings")].shape[0]
```

```python
# Toss winner is the match winner:
toss_win=df[df["TossWinner"]==df["WinningTeam"]].shape[0]
(toss_win/df.shape[0])*100
```
