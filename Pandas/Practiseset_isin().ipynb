```python
import pandas as pd

df = pd.DataFrame({
    "match_id": [1,1,1,2,2,3,3,3,3],
    "player": ["Kohli","Smith","Root","Kohli","Warner",
               "Smith","Root","Kohli","Buttler"],
    "runs": [45,60,12,80,10,100,5,23,44],
    "team": ["IND","AUS","ENG","IND","AUS",
             "AUS","ENG","IND","ENG"]
})

print(df)
```

“Find the total runs scored by players who have ever scored ≥ 50 runs in any match, but only in matches where the opponent team is also in a shortlist (['AUS','ENG']).”

```python
opponent_team=~df['team'].isin(['AUS','ENG'])
opponent_team
```

```python
players_fifty=df[df['runs']>50]['player'].unique()
players_fifty
```

```python
players=df['player'].isin(players_fifty)
players
```

```python
df[opponent_team & players]
```

You have a df of student grades with columns ['name', 'subject', 'grade'].
Find all rows where subject is either "Math" or "Science".

```python
df = pd.DataFrame({
    "name": ["Alice", "Bob", "Charlie", "David", "Eva"],
    "subject": ["Math", "English", "Science", "History", "Math"],
    "grade": [85, 70, 90, 65, 95]
})
print(df)
```

```python
df[df['subject'].isin(['Math','Science'])]
```

Given a df with sales data ['product', 'region', 'sales']:Extract all rows where region is not in ['North', 'South'].Now, from those rows, keep only the products "A" or "B".

```python
df = pd.DataFrame({
    "product": ["A", "B", "C", "D", "A", "B"],
    "region": ["North", "South", "East", "West", "South", "East"],
    "sales": [200, 150, 300, 400, 250, 350]
})
print(df)
```

```python
new_df=df[~df['region'].isin(['North', 'South'])]
new_df
```

```python
new_df[new_df['product'].isin(['B','A'])]
```
