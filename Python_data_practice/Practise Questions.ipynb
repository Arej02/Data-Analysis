### 1. List Comprehensions:

a. Create a list of the squares of numbers from 1 to 20.

b. Given a list of words, return a new list containing only words with more than 5 letters.

d. From a list of daily temperatures (in °C), create a new list containing only temperatures converted to °F, but only for days above 25°C.

e. Given a list of tuples (country, population_in_millions), create a new list of country names where the population is over 50 million.

### 2. Lambda Functions

a. Use a lambda to sort a list of tuples (name, age) by age.

b. Given a list of strings, use map with a lambda to convert them to uppercase.

c. Use filter with a lambda to return only even numbers from a list.

d. Given a list of dictionaries with keys {'city': str, 'temperature': float}, use sorted() and a lambda to sort by temperature in descending order.

e. Given a list of (student_name, score) tuples, use map and a lambda to convert each score to a percentage (out of 100).

```python
people = [
    ("Alice", 29),
    ("Bob", 34),
    ("Charlie", 22),
    ("Diana", 31),
    ("Ethan", 25),
    ("Fiona", 29),
    ("George", 40),
    ("Hannah", 34)
]
```

```python
#a:
sorted(people,key=lambda x:x[1])
```

```python
#b:
list(map(lambda x:x[0].upper(),people))
```

### 3. Membership Operator (in, not in)


a. Check if the number 42 exists in a given list.

b. From a list of names, create a new list excluding "John".

c. Check if all elements in one list exist in another list.

d. Given a Pandas column "Country", filter the DataFrame to only rows where "Country" is in a given list of target countries.

e. Using NumPy, check which elements in array A are also in array B.

f. Given a sentence, return all words that are not in a list of stopwords.



```python
#a:
l1=[12,45,56,32,56]
print(42 in l1)

l2=[12,45,56,32,56,42]
print(42 in l2)
```

```python
#b:
people2 = ["Alice","Bob", "Charlie", "Diana","Ethan","Fiona", "George","Hannah","John"]

people3=[name for name in people2 if name!='John']
people3
```
