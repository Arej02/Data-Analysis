### Basic Operations:

```python
titanic['Name'].str.lower() #str is the string accessor and lower will lowercase all the characters
```

```python
titanic['Name'].str.title() #Only make the title capital
```

```python
titanic['Name'].str.len().min() #len() finds the length of the character and min gives you the minmum length
```

```python
titanic['Name'][titanic['Name'].str.len().min()]
```

```python
name="     Arya   "
print(name)
print(name.strip()) #Strip will just remove the trailing spaces
```

### More Advanced Vectorization:

```python
#Seperate Name into three different columns(Title,First Name and Last Name) using split and get
titanic['Name']
```

```python
titanic['last_name']=titanic['Name'].str.split(',').str.get(0)
```

```python
titanic[["Title","First_Name"]]=titanic['Name'].str.split(',').str.get(1).str.strip().str.split('.',n=1,expand=True)
titanic.sample(3)
```

```python
titanic["Title"].unique()
```

```python
titanic["Title"].replace({'Ms':'Miss','Mr':'Master'},regex=False)
```

```python
titanic[titanic["Name"].str.startswith('A')]
```

### Regular Expression

```python
titanic[titanic['Name'].str.contains('Rosen')]
```

```python
titanic[titanic['Name'].str.contains('^[aeiouAEIOU].+[aeiouAEIOU]$')]
```
